# 1 - Operaciones con números complejos 

Un número complejo es un elemento de un sistema numérico que extiende a los números reales. La extensión se hace a través de la unidad imaginaria $i$ que por definición cumple con $i^2=-1$. No existe ningún número real que cumpla con esta condición, por lo tanto se lo conoce como un número imaginario. 

Un número commplejo $z$ se compone de una parte real  y una parte imaginaria como 

$$
z= x + i y,
$$

en donde $x$ y $y$ son números reales. Al número $x$ se lo conoce como la parte real de $z$ y a $y$ como su parte imaginaria. Esto se escribe como ${\rm Re}(z)=x$ e ${\rm Im}(z)=y$. Consecuentemente, un número complejo se lo puede representar en un plano de dos dimensiones llamado el "plano complejo". En el plano complejo por convención se grafica la parte real en el eje horizontal y la parte imaginaria en el eje vetical. Un ejemplo de esta representación es:

## Suma y producto

Sean $z = a + bi$ y $w = c + di$, con $a, b, c, d \in \mathbb{R}$.

**Suma.**
Se suma parte real con parte real y parte imaginaria con parte imaginaria. Por ejemplo:

$$z + w = (a + c) + (b + d)\,i .$$

Es igual a la suma de vectores en el plano. Si se dibuja $z$ y $w$ como flechas desde el origen, $z + w$ es la diagonal del paralelogramo.
La resta se hace de manera análoga. 

**Producto.** 
Se multiplica como en álgebra (conmutativa) estándar y al final se usa $i^2$. Un ejemplo:

$$zw = (a + bi)(c + di) = ac + adi + bci + bd\,i^2 = (ac - bd) + (ad + bc)\,i$$

El producto **no** es componente a componente. La parte real de $zw$ no es $ac$, porque el término $bd\,i^2$ se vuelve real y termina
restándose. El producto genera un reescalamiento y rotación en el plano, lo cual quedará más claro en la siguiente sección.

En python la unidad imaginaria es $i=$ 1j. Verifica lo aprendido en los siguientes ejercicios:

## Números complejos como vectores

Un número complejo se  puede tratar como un vector de dos
entradas, como:

$$z = a + ib \qquad \longleftrightarrow \qquad \vec{z} = \begin{pmatrix} a \\ b \end{pmatrix}.$$

Escrito así, la suma de dos complejos es simplemente la suma de dos vectores.

Para escribir el producto de dos números complejos con vectores se puede utilizar una matriz como ayuda.
Dicha matriz actua sobr el vector de manera que representa el producto. 
Por ejemplo, multiplicar por $w = c + id$ es aplicar


$$zw \qquad \longleftrightarrow \qquad
\begin{pmatrix} c & -d \\ d & c \end{pmatrix}
\begin{pmatrix} a \\ b \end{pmatrix}
= \begin{pmatrix} ac - bd \\ ad + bc \end{pmatrix},$$

cuyas dos entradas son la parte real y la parte imaginaria del número $zw$.

En numpy un vector columna se declara con `np.array([[a], [b]])` y el producto de una matriz por un
vector se hace con `@`. Nótese que `@` no es lo mismo que `*`, que multiplica entrada por entrada.

## Magnitud, argumento y forma polar

Existe otra manerra de representar un número complejo. Como cualquier punto en un plano, también se lo puede posicionar con coordenadas polares, es decir con su distancia al origin y su ángulo desde el eje horizontal. 

**Magnitud.**
La magnitud o módulo de $z = a+ ib$ es la distancia que hay deste el punto $z$ hasta el origen. Se calcula como:

$$r = |z| = \sqrt{a^2 + b^2} .$$

Al ser una distancia siempr se tiene $r \geq 0$. 
El conjugado del un número complejo $z^*$ (a veces denotado también como $\bar{z}$) se define como 

$$z= a + ib, \qquad z^* = a - ib$$


Una forma conveniente de calcular la magnitud es mediante
$$|z|^2 = z z^* .$$

**Argumento.**
El argumento de $z = x + iy$ es el ángulo $\varphi$ que forma la flecha desde el orgen hasta el punto $z$ con el eje horizontal. Una manera de calcularlo es:

$$\tan\theta = \frac{b}{a} .$$

Gráficamente la magnitud y el ángulo se ven como:

**Forma polar.**
Cualquier número complejo $z=a + ib$ con magnitud $r$ y argumento $\varphi$ se puede escribir como 


$$z = r(\cos(\varphi) + i\,\sin(\varphi)) .$$

La pareja $(r, \varphi)$ al igual que $(a, b)$.

En Python la magnitud se calcula con `abs(z)` y el argumento con `cmath.phase(z)`.
La función `cmath.polar(z)` devuelve la pareja $(r, \varphi)$ y se puede usar `cmath.rect(r, theta)` para revertir el cambio.
Los ángulos siempre están en radianes. Para importar estas funciones se puede usar `import cmath, numpy as np`

### Ejercicio 3

Sea $z = -1 + i$.

1. Calcule $r = |z|$ y $\theta = \arg z$.
2. Escriba $z$ en forma polar.

````python
# definimos el complejo
z = -1 + 1j

# magnitud y fase
r = abs(z)
print(f"Magnitud: {r}")

r_fase = cmath.phase(z)
print(f"Fase: {r_fase}")

# conversion a polar
polar = cmath.polar(z)
print(f"polar: {polar}")

# regresar a cartesianas
original = cmath.rect(r, r_fase)
print(f"Inverso: {original}")

# el punto esta en el cuadrante 2
print(f"Resultado final de theta: {r_fase}")
````

### Ejercicio 4

Sea $z = 2 + i$, considera $z^2$.

Calcula las magnitudes y argumentos de $z$ y $z^2$. ¿Cómo se relaciona la magnitud y el argumento de $z$ con las de $z^2$?

````python

# numero original y al cuadrado
w = 2 + 1j
w_cuadrado = w**2

# magnitudes
mag = abs(w)
mag_c = abs(w_cuadrado)
print(f"mag 1: {mag}")
print(f"mag 2: {mag_c}")

# angulos
ang = cmath.phase(w)
ang_c = cmath.phase(w_cuadrado)
print(f"ang 1: {ang}")
print(f"ang 2: {ang_c}")

# pasar a polares
par_polar = cmath.polar(w)
par_polar_c = cmath.polar(w_cuadrado)
print(f"polar 1: {par_polar}")
print(f"polar 2: {par_polar_c}")

# volver a cartesianas
al_reves = cmath.rect(mag, ang)
al_reves_c = cmath.rect(mag_c, ang_c)
print(f"rect 1: {al_reves}")
print(f"rect 2: {al_reves_c}")

# ubicacion final de los angulos
print(f"Angulo del primer punto: {ang}")
print(f"Angulo del punto al cuadrado: {ang_c}")
````

## La fórmula de Euler

Para cualquier ángulo real $\theta$ se cumple

$$e^{i\theta} = \cos\theta + i\,\sin\theta .$$

Este de resultado viene de comparar las series de Taylor de $e^x$, $\cos x$ y $\sin x$. 

Comparando con la sección anterior, un número complejo $z = r(\cos\theta + i\,\mathrm{sen}\,\theta)$ se puede escribir como 

$$z = r\,e^{i\theta} .$$

El producto entre dos números complejos se puede calcular mucho más fácil usando la formula de Euler. Por ejemplo:

$$\big(r e^{i\theta}\big)\big(s e^{i\varphi}\big) = rs\,e^{i(\theta + \varphi)} .$$


Cuando $r = 1$ queda $e^{i\theta}$ solo, que siempre tiene magnitud $1$ y por lo
tanto vive sobre el círculo unitario. A un número de esta forma se le llama
fase, y multiplicar por él no cambia la magnitud de un número, solamente lo rota.

En Python la exponencial compleja es `cmath.exp(1j*theta)` para un solo número, o
`np.exp(1j*theta)` si `theta` es un arreglo de ángulos.
